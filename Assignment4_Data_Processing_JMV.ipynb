##**Task 1: Create a DataFrame**

This task creates a Pandas DataFrame containing names, ages, and cities. Data types are verified to ensure correctness.

```python
# Task 1: Create a DataFrame

import pandas as pd
import numpy as np
from functools import reduce

try:
    data = {
        "Name": ["Alice", "Bob", "Charlie", "David"],
        "Age": [25, 30, 35, 28],
        "City": ["New York", "San Francisco", "Los Angeles", "Chicago"]
    }

    df = pd.DataFrame(data)

    print("DataFrame:")
    print(df)

    print("\nData Types:")
    print(df.dtypes)

except Exception as e:
    print("Error:", e)
```

##**Task 3: Handling Null Values**
This task demonstrates how to identify and fill missing values using Pandas.

```python
# Task 3: Handling Null Values

try:
    df_null = pd.DataFrame({
        "Name": ["John", None, "Mary"],
        "Age": [25, np.nan, 30],
        "City": ["Toronto", "Ottawa", None]
    })

    print("Original DataFrame:")
    print(df_null)

    filled_df = df_null.fillna({
        "Name": "Unknown",
        "Age": 0,
        "City": "Not Provided"
    })

    print("\nFilled DataFrame:")
    print(filled_df)

    print("\nOther ways to handle missing values:")
    print("- dropna() removes rows with missing values")
    print("- fillna() replaces missing values")
    print("- interpolate() estimates missing numeric values")

except Exception as e:
    print("Error:", e)
```

#**Task 4: GroupBy and Describe**
Data is grouped by category and descriptive statistics are generated for the Value column.

```python
# Task 4: GroupBy and Describe

df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

try:
    result = df_group.groupby("Category")["Value"].describe()

    print("Grouped Statistics:")
    print(result)

    print("\nInterpretation:")
    print("Category A has 3 values with an average of 18.33.")
    print("Category B has 2 values with an average of 22.5.")
    print("Category C has one value, so mean equals 35.")

except Exception as e:
    print("Error:", e)
```

##**Task 5: Concatenation and Merging**
Two DataFrames are concatenated vertically and merged with another DataFrame horizontally.

```python
# Task 5: Concatenation and Merging

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

try:
    concatenated_df = pd.concat([df1, df2], ignore_index=True)

    merged_df = pd.concat(
        [concatenated_df.reset_index(drop=True),
         df3.reindex(range(len(concatenated_df)))],
        axis=1
    )

    print("Concatenated DataFrame:")
    print(concatenated_df)

    print("\nMerged DataFrame:")
    print(merged_df)

    print("\nShape:", merged_df.shape)

except Exception as e:
    print("Error:", e)
```

##**Task 6: Tuples and Sets**
This task demonstrates the difference between mutable and immutable data structures.

```python
# Task 6: Tuples and Sets

fruits = ("Apple", "Banana", "Orange")
numbers_set = {1, 2, 3, 4, 5}

print("Original Tuple:", fruits)
print("Original Set:", numbers_set)

try:
    fruits.append("Mango")
except AttributeError as e:
    print("\nCannot modify tuple:")
    print(e)

numbers_set.add(6)

print("\nUpdated Set:")
print(numbers_set)

print("\nExplanation:")
print("Tuples are immutable.")
print("Sets are mutable and allow adding/removing elements.")
```

##**Task 8: Functions and Lambda**
Comparison between a regular function and a lambda function.

```python
# Task 8: Functions and Lambda

def square(number):
    return number ** 2

square_lambda = lambda number: number ** 2

for num in [4, 8]:
    print(f"Regular Function: square({num}) = {square(num)}")
    print(f"Lambda Function: square({num}) = {square_lambda(num)}")
```

##**Task 9: Iterators and Generators**
Creating both an iterator class and a generator function.

```python
# Task 9: Iterators and Generators

class EvenIterator:
    def __init__(self):
        self.num = 0
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= 5:
            raise StopIteration

        value = self.num
        self.num += 2
        self.count += 1
        return value

print("Iterator Output:")
for num in EvenIterator():
    print(num)

def even_generator():
    for i in range(0, 10, 2):
        yield i

print("\nGenerator Output:")
for num in even_generator():
    print(num)
```

##**Task 10: Map, Reduce and Filter**
Using functional programming techniques on a list of numbers.

```python
# Task 10: Map, Reduce and Filter

numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x**2, numbers))

product = reduce(lambda x, y: x * y, numbers)

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Squared Numbers:", squared)
print("Product:", product)
print("Even Numbers:", even_numbers)
```

##**Task 11: Object-Oriented Programming**
Creating a Rectangle class with methods to calculate area and perimeter.

```python
# Task 11: Object-Oriented Programming

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect1 = Rectangle(10, 5)
rect2 = Rectangle(8, 4)

print("Rectangle 1")
print("Area:", rect1.area())
print("Perimeter:", rect1.perimeter())

print("\nRectangle 2")
print("Area:", rect2.area())
print("Perimeter:", rect2.perimeter())
```

##**Task 12: Pandas Data Analysis**
Analyzing employee salary data using Pandas.

```python
# Task 12: Pandas Data Analysis

df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

try:
    average_salary = df_employees.groupby("Department")["Salary"].mean()

    print("Average Salary by Department:")
    print(average_salary)

    high_salary = df_employees[df_employees["Salary"] > 60000]["Name"]

    print("\nEmployees with Salary > 60000:")
    print(high_salary.tolist())

    df_employees["Bonus"] = df_employees["Salary"] * 0.10

    print("\nUpdated Employee Data:")
    print(df_employees)

except Exception as e:
    print("Error:", e)
```
