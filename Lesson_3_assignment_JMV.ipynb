## **6. Functions and conditional statements and checking availability function**

A function is a reusable block of code. `check_availability()` checks whether a requested title exists in the current book list. The comparison ignores differences in uppercase and lowercase letters and extra spaces.

```python
def check_availability(book_name):
    """Return True if book_name is in the library list; otherwise return False."""

    # Remove extra spaces and compare titles without case sensitivity.
    requested_title = book_name.strip().casefold()

    # An empty title cannot match a valid book.
    if not requested_title:
        return False

    # Check whether any stored title matches the requested title.
    for book in books:
        if book.casefold() == requested_title:
            return True

    # Return False when no matching title is found.
    return False


# Test the function with a book that exists.
print("Is 'Python Basics' available?", check_availability("Python Basics"))

# Test the function with a title that does not exist.
print("Is 'Unknown Book' available?", check_availability("Unknown Book"))

# Test case-insensitive matching.
print("Is 'python basics' available?", check_availability("python basics"))
```

## **7. File I/O and saving and reading books from a text file**

File handling allows a program to store information after it stops running. The first function writes one book title per line. The second function reads the titles back and handles a missing file or other file-related errors.

```python
def save_books_to_file(filename):
    """Save the current book list to a UTF-8 text file."""

    try:
        # Open the file in write mode; this creates or replaces the file.
        with open(filename, "w", encoding="utf-8") as file:
            for book in books:
                file.write(book + "\\n")

        print(f"Successfully saved {len(books)} books to '{filename}'.")

    except OSError as error:
        # Report file-system errors such as permission or path problems.
        print(f"Could not save the books: {error}")


def read_books_from_file(filename):
    """Read book titles from a text file and return them as a list."""

    try:
        # Open the file in read mode and load non-empty lines.
        with open(filename, "r", encoding="utf-8") as file:
            loaded_books = [line.strip() for line in file if line.strip()]

        return loaded_books

    except FileNotFoundError:
        # Handle the case where the requested file does not exist.
        print(f"File not found: '{filename}'.")
        return []

    except OSError as error:
        # Handle other file-reading problems.
        print(f"Could not read the file: {error}")
        return []


# Save the current book list and then read it back.
text_filename = "books.txt"
save_books_to_file(text_filename)

books_from_file = read_books_from_file(text_filename)
print("Books read from the text file:")
for number, title in enumerate(books_from_file, start=1):
    print(f"{number}. {title}")
```

## **8. Pickle function serializing and loading the book list**

Pickle converts a Python object into a binary file so it can be stored and restored later. Only load Pickle files from trusted sources because loading an untrusted Pickle file can execute malicious code.

```python
def save_books_pickle(filename):
    """Serialize the current book list into a Pickle file."""

    try:
        # Open the file in binary-write mode and serialize the list.
        with open(filename, "wb") as file:
            pickle.dump(books, file)

        print(f"Successfully saved {len(books)} books to '{filename}'.")

    except (OSError, pickle.PicklingError) as error:
        print(f"Could not save the Pickle file: {error}")


def load_books_pickle(filename):
    """Load and return a book list from a trusted Pickle file."""

    try:
        # Open the Pickle file in binary-read mode.
        with open(filename, "rb") as file:
            loaded_books = pickle.load(file)

        # Confirm that the loaded object has the expected list structure.
        if not isinstance(loaded_books, list):
            print("The Pickle file does not contain a list.")
            return []

        return loaded_books

    except FileNotFoundError:
        print(f"Pickle file not found: '{filename}'.")
        return []

    except (OSError, pickle.UnpicklingError, EOFError) as error:
        print(f"Could not load the Pickle file: {error}")
        return []


# Save the list and load it back to verify serialization.
pickle_filename = "books.pkl"
save_books_pickle(pickle_filename)

books_from_pickle = load_books_pickle(pickle_filename)
print("Books loaded from Pickle:", books_from_pickle)
```

## **9. SQLite database interaction**

SQLite is a lightweight relational database stored in a local file. This section creates a `books` table and inserts book records into it.

The insert function validates its inputs and uses SQL placeholders (`?`) rather than combining values directly into the SQL statement. This is safer and helps prevent SQL injection.

```python
def create_books_table():
    """Create the books table if it does not already exist."""

    try:
        # Connect to the database file; SQLite creates it if necessary.
        with sqlite3.connect("library.db") as connection:
            cursor = connection.cursor()

            # Create a table with an ID, title, and available quantity.
            cursor.execute(
                """
                CREATE TABLE IF NOT EXISTS books (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    title TEXT NOT NULL UNIQUE,
                    quantity INTEGER NOT NULL CHECK (quantity >= 0)
                )
                """
            )

        print("Books table is ready.")

    except sqlite3.Error as error:
        print(f"Database table creation failed: {error}")


def add_book_to_db(title, quantity):
    """Insert a book and its quantity into the SQLite database."""

    # Validate the title before connecting to the database.
    if not isinstance(title, str) or not title.strip():
        print("Book title must be a non-empty string.")
        return False

    # Reject negative quantities and non-integer quantities.
    if not isinstance(quantity, int) or quantity < 0:
        print("Quantity must be a non-negative integer.")
        return False

    try:
        # The context manager commits on success and rolls back on error.
        with sqlite3.connect("library.db") as connection:
            cursor = connection.cursor()

            # Use parameterized SQL to safely insert user-provided values.
            cursor.execute(
                "INSERT INTO books (title, quantity) VALUES (?, ?)",
                (title.strip(), quantity)
            )

        print(f"Added '{title.strip()}' to the database.")
        return True

    except sqlite3.IntegrityError:
        # The UNIQUE constraint prevents duplicate book titles.
        print(f"'{title.strip()}' is already in the database.")
        return False

    except sqlite3.Error as error:
        print(f"Could not add the book to the database: {error}")
        return False


def display_books_from_db():
    """Retrieve and display all records stored in the books table."""

    try:
        with sqlite3.connect("library.db") as connection:
            cursor = connection.cursor()
            cursor.execute("SELECT id, title, quantity FROM books ORDER BY id")
            records = cursor.fetchall()

        print("Books stored in the SQLite database:")
        if not records:
            print("No books have been added yet.")
        else:
            for record in records:
                print(f"ID: {record[0]} | Title: {record[1]} | Quantity: {record[2]}")

    except sqlite3.Error as error:
        print(f"Could not retrieve database records: {error}")


# Create the table before inserting any records.
create_books_table()

# Insert two sample records for testing.
add_book_to_db("Python for Beginners", 10)
add_book_to_db("Data Analytics Fundamentals", 6)

# Display the records to verify that insertion worked.
display_books_from_db()
```

## **10. Web API interaction and GitHub repositories**

An API allows programs to request data from another service. This section uses the GitHub Search API to retrieve the names of the five most-starred public repositories written in Python.

The request includes a timeout and checks the HTTP response so that network or API errors are handled clearly.

```python
def get_python_repos():
    """Return the names of up to five popular public Python repositories."""

    # GitHub endpoint for searching public repositories.
    url = "https://api.github.com/search/repositories"

    # Search for Python repositories and sort them by star count.
    params = {
        "q": "language:python",
        "sort": "stars",
        "order": "desc",
        "per_page": 5
    }

    try:
        # Send the request with a timeout to avoid waiting indefinitely.
        response = requests.get(url, params=params, timeout=15)

        # Raise an exception for unsuccessful HTTP status codes.
        response.raise_for_status()

        # Convert the JSON response into a Python dictionary.
        data = response.json()

        # Extract repository names from the returned items.
        return [repository["name"] for repository in data.get("items", [])[:5]]

    except requests.exceptions.Timeout:
        print("The GitHub API request timed out.")
        return []

    except requests.exceptions.HTTPError as error:
        print(f"GitHub returned an HTTP error: {error}")
        return []

    except requests.exceptions.RequestException as error:
        print(f"Could not connect to the GitHub API: {error}")
        return []

    except (ValueError, KeyError) as error:
        print(f"Could not process the GitHub API response: {error}")
        return []


# Call the API function and display the repository names.
top_python_repositories = get_python_repos()

print("Top Python repositories on GitHub:")
if top_python_repositories:
    for position, repository_name in enumerate(top_python_repositories, start=1):
        print(f"{position}. {repository_name}")
else:
    print("No repository names were returned. Check the connection and try again.")
```
