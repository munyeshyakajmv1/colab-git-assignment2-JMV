### 7.2 Authenticate securely and push to GitHub

Cloning normally creates the `origin` remote automatically. Do not run `git remote add origin` again if `git remote -v` already displays it.

The following cell requests the PAT through a hidden prompt and uses a temporary ask-pass helper. The token is not inserted into a Git URL or notebook source. It is held in the current Colab runtime only; a runtime restart clears it.

Use a fine-grained token limited to this repository with **Contents: Read and write** permission.

```python
# getpass hides the token as it is typed, so it is not displayed in cell output.
from getpass import getpass
import os
import subprocess
from pathlib import Path

# Ask for the GitHub username and PAT privately.
github_user = input("GitHub username: ")
github_pat = getpass("GitHub PAT (input is hidden): ")

# Create a temporary script that supplies credentials when Git asks for them.
askpass_path = Path("/tmp/git-askpass.sh")
askpass_path.write_text(
    '#!/bin/sh\n'
    'case "$1" in\n'
    '  *Username*) echo "$GIT_USERNAME" ;;\n'
    '  *Password*) echo "$GIT_TOKEN" ;;\n'
    '  *) echo "" ;;\n'
    'esac\n'
)
askpass_path.chmod(0o700)

# Pass credentials through environment variables, not through the Git URL.
git_env = os.environ.copy()
git_env["GIT_ASKPASS"] = str(askpass_path)
git_env["GIT_USERNAME"] = github_user
git_env["GIT_TOKEN"] = github_pat
git_env["GIT_TERMINAL_PROMPT"] = "0"

# Verify that the repository has an origin remote.
remote_check = subprocess.run(
    ["git", "remote", "get-url", "origin"],
    capture_output=True,
    text=True
)
if remote_check.returncode != 0:
    raise RuntimeError("The origin remote is missing. Check the repository setup.")

# Push the main branch to GitHub.
push_result = subprocess.run(
    ["git", "push", "-u", "origin", "main"],
    env=git_env,
    text=True
)

# Remove the token from the Python variable and environment after the push.
github_pat = None
git_env.pop("GIT_TOKEN", None)

if push_result.returncode == 0:
    print("Push completed successfully.")
else:
    raise RuntimeError(
        "Push failed. Check your username, token permissions, repository URL, and branch name."
    )
```

### 8.1 Add and test a division function

The function checks whether the denominator is zero before dividing. This prevents a division-by-zero error and returns a clear message for that case.

```python
# Define a function that divides a numerator by a denominator.
def divide_numbers(a, b):
    # Check for zero before performing division.
    if b != 0:
        return a / b
    else:
        # Return a helpful message instead of attempting division by zero.
        return "Cannot divide by zero"


# Test normal division; the expected result is 5.0.
print("Division result:", divide_numbers(10, 2))

# Test the zero-denominator condition.
print("Zero-division test:", divide_numbers(10, 0))
```
